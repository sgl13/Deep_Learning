```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
#print the cv2 version
print("open CV version:",cv2.__version__)
#cv2 :- is a python package for opencv
```

```python
#what is a image?
#an image is a matrix of pixel values
#for a color image --> Height * width * channels(3)
#what are this 3 channels?
#channel1 - 1 --> Red
#channel2 - 2 --> Green
#channel3 - 3 --> Blue
#but opencv use it as a BGR
```

```python
#create a color image
n = np.zeros((500,700,3),dtype=np.uint8)
print(n)
print(n.shape)
```

```python
#create a rectangle
cv2.rectangle(
    n,
    (50,50), #starting point
    (250,250), #ending point
    (0,250,0), #BGR color
    -1
)
```

```python
#draw a circle with red color
cv2.circle(
    n,
    (550,350),
    90,
    (0,0,250),
    -1
)
```

```python
#opencv scores the image in BGR
#when plot the image by using the matplotlib except the image in RGB
#therefore convert the BGR image to RGB
rgb_image = cv2.cvtColor(n, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10,7))
plt.imshow(rgb_image)
plt.show()
#Q)why we use "cvtColor"
```

```python
#check the image channels
blue = image[:,:,0]
green = image[:,:,1]
red = image[:,:,2]

print(blue.shape)
print(green.shape)
print(red.shape)
```

```python
plt.figure(figsize=(15,4))
plt.subplot(1,3,1)
plt.imshow(blue,cmap='gray')
plt.title("blue channel")
plt.show()
```

```python
plt.figure(figsize=(15,4))
plt.subplot(1,3,2)
plt.imshow(green)
plt.title("green channel")
plt.show()
```

```python
plt.figure(figsize=(15,4))
plt.subplot(1,3,3)
plt.imshow(red)
plt.title("red channel")
plt.show()
```

```python
#resize of the image
resize = cv2.resize(image,(800,600)) #cv2.resize(image,(width,height))
print(image.shape)
print(resize.shape)
```

```python
#image cropping
crop = image[100:400,200:600]
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
#rotate the image
h,w = image.shape[:2]
#print(h,w)
center = (w//2,h//2)
#to rotate the image use matrix
matrix = cv2.getRotationMatrix2D(
    center,
    45, # rotation angle
    1.0
)
rotated = cv2.warpAffine(
    image,
    matrix,
    (w,h)
)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
# image blurr
blur_image = cv2.GaussianBlur(
    image,
    (7,7),
    0
    )
plt.imshow(cv2.cvtColor(blur_image, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
#convert to gray
gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
    )
plt.imshow(gray,cmap='gray')
plt.show()
```
